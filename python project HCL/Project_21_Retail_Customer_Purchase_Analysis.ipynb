```python
# Importing the libraries required for data loading, analysis, statistics, and charts.
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

DATA_FILE = Path("21_retail_customer_orders.csv")
if not DATA_FILE.exists():
    raise FileNotFoundError(
        f"Could not find {DATA_FILE}. Keep the CSV in the same folder as this notebook."
    )

# Load the source data without changing it.
raw_df = pd.read_csv(DATA_FILE)

# Keep a small output folder for cleaned data and analysis tables.
OUTPUT_DIR = Path("project21_outputs")
OUTPUT_DIR.mkdir(exist_ok=True)
#  original dataset before making any cleaning changes.

print(f"Loaded dataset: {DATA_FILE.resolve()}")
print(f"Original rows: {len(raw_df):,}")
print("Dataset dimensions (rows, columns):", raw_df.shape)
print("\nColumn names:")
print(raw_df.columns.tolist())
print("\nData types:")
print(raw_df.dtypes)
print("\nDataFrame information:")
raw_df.info()

#a few records to understand the structure and values.
display(raw_df.head(10))
```

## Question 2. Identify missing values and duplicate Order_ID records.

A repeated Order_ID is counted separately from a fully duplicated row. This helps distinguish repeated order identifiers from exact copied records.

```python
#  missing values in every column.
missing_summary = raw_df.isna().sum().rename("Missing_Values").to_frame()
display(missing_summary)

# how many Order_ID values occur more than once.
duplicate_id_mask = raw_df.duplicated(subset="Order_ID", keep=False)
duplicate_ids = sorted(raw_df.loc[duplicate_id_mask, "Order_ID"].unique().tolist())
extra_duplicate_rows = raw_df.duplicated(subset="Order_ID", keep="first").sum()

# Check whether the repeated Order_ID records are exact row-for-row copies.
fully_duplicated_rows = raw_df.duplicated().sum()
print(f"Distinct Order_ID values with repetitions: {len(duplicate_ids)}")
print(f"Rows belonging to repeated Order_ID groups: {int(duplicate_id_mask.sum())}")
print(f"Extra rows beyond the first copy of each Order_ID: {int(extra_duplicate_rows)}")
print(f"Fully duplicated rows across all columns: {int(fully_duplicated_rows)}")
print("Repeated Order_ID values:\n", duplicate_ids)

# Show all rows for the repeated order identifiers for auditability.
if duplicate_ids:
    display(raw_df.loc[duplicate_id_mask].sort_values("Order_ID"))
```

## Question 3. Handle missing Quantity or Unit_Price values using a justified strategy.


```python
# separate copy so the raw source DataFrame remains unchanged.
clean_df = raw_df.drop_duplicates().copy()

# missing counts for a before/after comparison.
missing_quantity_before = int(clean_df["Quantity"].isna().sum())
missing_price_before = int(clean_df["Unit_Price"].isna().sum())

# Fill Quantity using its median, then make sure quantity remains a whole number.
quantity_median = clean_df["Quantity"].median()
clean_df["Quantity"] = clean_df["Quantity"].fillna(quantity_median).round().astype(int)

# Calculate each product's median observed price for product-specific imputation.
product_price_medians = clean_df.groupby("Product_Name")["Unit_Price"].median()
overall_price_median = clean_df["Unit_Price"].median()
product_price_fill = clean_df["Product_Name"].map(product_price_medians)
clean_df["Unit_Price"] = clean_df["Unit_Price"].fillna(product_price_fill)
clean_df["Unit_Price"] = clean_df["Unit_Price"].fillna(overall_price_median)

# Confirm that the selected fields no longer contain missing values.
print(f"Exact duplicate rows removed: {len(raw_df) - len(clean_df)}")
print(f"Missing Quantity values filled: {missing_quantity_before}")
print(f"Missing Unit_Price values filled: {missing_price_before}")
print("Quantity median used:", quantity_median)
print("Overall Unit_Price median (fallback only):", overall_price_median)
print("Remaining missing Quantity values:", int(clean_df["Quantity"].isna().sum()))
print("Remaining missing Unit_Price values:", int(clean_df["Unit_Price"].isna().sum()))
print("Cleaned dimensions:", clean_df.shape)
```

## Question 4. Convert Order_Date to datetime and extract the month, weekday, and year.

```python
# break down the order date; invalid values would become NaT and are checked below.
clean_df["Order_Date"] = pd.to_datetime(clean_df["Order_Date"], errors="coerce")
invalid_dates = int(clean_df["Order_Date"].isna().sum())
if invalid_dates:
    raise ValueError(f"Found {invalid_dates} invalid Order_Date values; inspect before analysis.")

# Extract the requested calendar features into separate columns.
clean_df["Month"] = clean_df["Order_Date"].dt.month
clean_df["Month_Name"] = clean_df["Order_Date"].dt.month_name()
clean_df["Weekday"] = clean_df["Order_Date"].dt.day_name()
clean_df["Year"] = clean_df["Order_Date"].dt.year

print("Invalid dates after conversion:", invalid_dates)
print("Order date range:", clean_df["Order_Date"].min().date(), "to", clean_df["Order_Date"].max().date())
display(clean_df[["Order_ID", "Order_Date", "Month", "Month_Name", "Weekday", "Year"]].head())
```

## Question 5. Calculate total orders, total units sold, and total sales revenue.



```python
# Each Order_ID represents one order after exact duplicate rows are removed.
total_orders = clean_df["Order_ID"].nunique()
total_units_sold = int(clean_df["Quantity"].sum())
total_sales_revenue = float((clean_df["Quantity"] * clean_df["Unit_Price"]).sum())

summary_q5 = pd.DataFrame({
    "Metric": ["Total orders", "Total units sold", "Total sales revenue"],
    "Value": [total_orders, total_units_sold, total_sales_revenue],
})
display(summary_q5)
```

## Question 6. Create a row-wise Order_Value column by multiplying Quantity by Unit_Price.

```python
# Calculate the revenue contributed by each order row.
clean_df["Order_Value"] = clean_df["Quantity"] * clean_df["Unit_Price"]

# Verify that the new column agrees with the calculation in Question 5.
assert np.isclose(clean_df["Order_Value"].sum(), total_sales_revenue)
print("Order_Value column created successfully.")
display(clean_df[["Order_ID", "Quantity", "Unit_Price", "Order_Value"]].head())
```

## Question 7. Find the 10 products with the highest sales revenue.

The dataset contains fewer than 10 distinct product names, so the table displays every available product when fewer than 10 exist.

```python
# Rank products by their total Order_Value and total units sold.
top_products = (
    clean_df.groupby("Product_Name", as_index=False)
    .agg(
        Total_Quantity=("Quantity", "sum"),
        Total_Revenue=("Order_Value", "sum"),
        Order_Count=("Order_ID", "nunique"),
    )
    .sort_values("Total_Revenue", ascending=False)
    .head(10)
    .reset_index(drop=True)
)
top_products.insert(0, "Revenue_Rank", np.arange(1, len(top_products) + 1))
display(top_products.round({"Total_Revenue": 2}))

# Export the ranking so it can be reused outside the notebook.
top_products.to_csv(OUTPUT_DIR / "top_10_products_by_revenue.csv", index=False)
```

## Question 8. Rank product categories by total quantity sold and total revenue.

```python
# Aggregate units and revenue for each category, then calculate both rankings.
category_summary = (
    clean_df.groupby("Category", as_index=False)
    .agg(
        Total_Quantity=("Quantity", "sum"),
        Total_Revenue=("Order_Value", "sum"),
        Order_Count=("Order_ID", "nunique"),
    )
)
category_summary["Quantity_Rank"] = category_summary["Total_Quantity"].rank(
    method="dense", ascending=False
).astype(int)
category_summary["Revenue_Rank"] = category_summary["Total_Revenue"].rank(
    method="dense", ascending=False
).astype(int)
category_summary = category_summary.sort_values("Revenue_Rank").reset_index(drop=True)
display(category_summary.round({"Total_Revenue": 2}))
category_summary.to_csv(OUTPUT_DIR / "category_wise_sales_summary.csv", index=False)
```

## Question 9. Compare average order value across payment methods.

Because the dataset has one row per order after duplicate removal, the mean of `Order_Value` within each payment method is the average order value for that method.

```python
# Compare payment methods using order count, average order value, and total revenue.
payment_summary = (
    clean_df.groupby("Payment_Method", as_index=False)
    .agg(
        Orders=("Order_ID", "nunique"),
        Average_Order_Value=("Order_Value", "mean"),
        Total_Revenue=("Order_Value", "sum"),
    )
    .sort_values("Average_Order_Value", ascending=False)
    .reset_index(drop=True)
)
display(payment_summary.round(2))
payment_summary.to_csv(OUTPUT_DIR / "payment_method_comparison.csv", index=False)
```

## Question 10. Identify the busiest shopping days and the peak sales month.

To avoid ambiguity, this section reports (1) the busiest weekday by order count, (2) the busiest individual calendar dates, (3) the highest-revenue year-month, and (4) the highest-revenue calendar month when years are combined.

```python
import pandas as pd
from IPython.display import display

# Convert Order_Date into datetime format
clean_df["Order_Date"] = pd.to_datetime(
    clean_df["Order_Date"],
    errors="coerce"
)

# Extract weekday and month
clean_df["Weekday"] = clean_df["Order_Date"].dt.day_name()
clean_df["Year_Month"] = clean_df["Order_Date"].dt.to_period("M").astype(str)

# Calculate orders and revenue by weekday
weekday_summary = (
    clean_df.groupby("Weekday")
    .agg(
        Orders=("Order_ID", "nunique"),
        Revenue=("Order_Value", "sum")
    )
    .reindex([
        "Monday", "Tuesday", "Wednesday",
        "Thursday", "Friday", "Saturday", "Sunday"
    ])
    .reset_index()
)

print("Orders and Revenue by Weekday:")
display(weekday_summary)

# Calculate monthly sales
monthly_summary = (
    clean_df.groupby("Year_Month")
    .agg(
        Orders=("Order_ID", "nunique"),
        Revenue=("Order_Value", "sum")
    )
    .reset_index()
    .sort_values("Revenue", ascending=False)
)

print("Top 10 Months by Revenue:")
display(monthly_summary.head(10))

# Identify busiest calendar dates
daily_summary = (
    clean_df.groupby(clean_df["Order_Date"].dt.date)
    .agg(
        Orders=("Order_ID", "nunique"),
        Revenue=("Order_Value", "sum")
    )
    .reset_index()
    .rename(columns={"Order_Date": "Order_Date"})
    .sort_values("Orders", ascending=False)
)

print("Top 10 Busiest Shopping Dates:")
display(daily_summary.head(10))
```

## Question 14. Identify unusually high order values using the IQR method and create Low, Medium, and High order-value bands.

**Outlier rule:** Values below `Q1 - 1.5 × IQR` or above `Q3 + 1.5 × IQR` are flagged as IQR outliers.  
**Band rule:** Low, Medium, and High use the 33⅓th and 66⅔th percentiles as cutoffs. This creates data-relative bands and does not imply a business-approved monetary threshold.

```python
# Calculate the first and third quartiles and derive the IQR fences.
q1, q3 = np.percentile(order_values, [25, 75])
iqr = q3 - q1
lower_fence = q1 - 1.5 * iqr
upper_fence = q3 + 1.5 * iqr

# Select unusually low or high order values using the standard IQR rule.
outlier_mask = (clean_df["Order_Value"] < lower_fence) | (clean_df["Order_Value"] > upper_fence)
outlier_orders = clean_df.loc[
    outlier_mask,
    ["Order_ID", "Order_Date", "Product_Name", "Category", "Quantity", "Unit_Price", "Order_Value"],
].sort_values("Order_Value", ascending=False)

print(f"Q1: {q1:,.2f}")
print(f"Q3: {q3:,.2f}")
print(f"IQR: {iqr:,.2f}")
print(f"Lower fence: {lower_fence:,.2f}")
print(f"Upper fence: {upper_fence:,.2f}")
print(f"Number of IQR outlier orders: {len(outlier_orders)}")
display(outlier_orders.round({"Unit_Price": 2, "Order_Value": 2}))
outlier_orders.to_csv(OUTPUT_DIR / "iqr_outlier_orders.csv", index=False)

# Use percentile-based cutoffs to create reproducible order-value bands.
low_cutoff, high_cutoff = np.percentile(order_values, [100 / 3, 200 / 3])
clean_df["Order_Value_Band"] = np.select(
    [clean_df["Order_Value"] <= low_cutoff, clean_df["Order_Value"] <= high_cutoff],
    ["Low", "Medium"],
    default="High",
)
order_band_summary = (
    clean_df.groupby("Order_Value_Band", as_index=False)
    .agg(Orders=("Order_ID", "nunique"), Average_Order_Value=("Order_Value", "mean"))
)
print(f"Band cutoffs: Low <= {low_cutoff:,.2f}; Medium <= {high_cutoff:,.2f}; High > {high_cutoff:,.2f}")
display(order_band_summary.round(2))
```
